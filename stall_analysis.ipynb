## Introduction

This analysis examines two weeks of sales and stock records from Wanjiku's fresh-produce stall at Gikomba Market. The objective is to understand the stall's financial performance, identify sales trends, evaluate stock levels and calculate product gross margins. The findings will help Wanjiku make informed decisions about restocking and business operations.

## Objectives

1. Analyse total, average, highest and lowest daily sales.
2. Compare sales performance between the two weeks.
3. Classify days according to sales performance.
4. Compare weekday and weekend sales.
5. Identify products requiring restocking and calculate gross margins.
6. Generate a formatted stall performance report.
7. Provide data-driven business recommendations.


```python

daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```


## Task 1: Overall Sales Performance

In this task, I will calculate the total and average sales over the 14-day period. I will also identify the highest and lowest sales recorded and determine the corresponding day and week.


```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

print("Total Sales:", total_sales)
print("Average Daily Sales:", average_sales)
print("Highest Sales:", highest_sales)
print("Lowest Sales:", lowest_sales)

print(daily_sales.index(highest_sales))
print(daily_sales.index(lowest_sales))
```

```python
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
worst_day = days[worst_index]

best_week = 1 if best_index < 7 else 2
worst_week = 1 if worst_index < 7 else 2

print("Best Day:", best_day, "- Week", best_week)
print("Worst Day:", worst_day, "- Week", worst_week)
```


## Task 2: Weekly Sales and Growth Analysis

In this task, I will divide the 14 days of sales into two weeks using list slicing. I will create a reusable function to calculate each week's total sales and determine the percentage growth between the two weeks.


```python
def week_totals(sales):
    """ Calculate and return total sales for each of the two weeks."""
    week1 = sales[0:7]
    week2 = sales[7:14]

    week1_total = sum(week1)
    week2_total = sum(week2)
    
    return week1_total, week2_total

week1_total, week2_total = week_totals(daily_sales)

print("Week 1:", week1_total)
print("Week 2:", week2_total)

growth = ((week2_total - week1_total) / week1_total) * 100
print("Week-on-week growth:", growth, "%")
```

 
## Task 3: Daily Sales Classification

In this task, I will classify each day's sales as Strong, Normal or Weak based on the overall average sales. I will then count the number of days in each category using a dictionary.


```python
def classify_day(amount, average):
    """Classify daily sales based on the average sales."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

print(classify_day(5000, average_sales))
print(classify_day(3300, average_sales))
print(classify_day(2000, average_sales))
```

```python
counts = { "Strong" : 0,
           "Normal" : 0,
           "Weak": 0

         }

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    counts[category] += 1

print(counts)
```

```python
for i in range(len(daily_sales)):
    category = classify_day(daily_sales[i], average_sales)
    print(days[i], "-", daily_sales[i], "-", category)
```


## Task 4: Weekday vs Weekend Sales Analysis

In this task, I will compare average sales on weekdays (Monday to Friday) against weekends (Saturday and Sunday). I will calculate the percentage difference to identify which period generates higher daily sales.


```python
week1 = daily_sales[0:7]

weekdays_week1 = week1[0:5]
weekends_week1 = week1[5:7]

week2 = daily_sales[7:14]

weekdays_week2 = week2[0:5]
weekends_week2 = week2[5:7]

weekday_sales = weekdays_week1 + weekdays_week2
weekend_sales = weekends_week1 + weekends_week2

avg_weekday_sales = sum(weekday_sales) / len(weekday_sales)
avg_weekend_sales = sum(weekend_sales) / len(weekend_sales)

print("Average weekday sales:", avg_weekday_sales)
print("Average weekend sales:", avg_weekend_sales)

percentage_difference = ((avg_weekend_sales - avg_weekday_sales) / avg_weekday_sales) * 100
print(percentage_difference)
```

## Task 5: Stock Management and Gross Margin Analysis

In this task, I will analyse the current stock levels to identify products that need restocking based on their reorder levels. I will also calculate the gross margin percentage for each product to assess its profitability and support better purchasing and pricing decisions.

```python
def reorder_list(stock):
    """Return products that need restocking"""
    new_item_list = []
    
    for product in stock:
        quantity = stock[product]["qty"]
        reorder_level = stock[product]["reorder"]
    
        if quantity <= reorder_level:
            new_item_list.append(product)

    return new_item_list

print("Products to reorder:", reorder_list(stock))

for product in stock:
    cost = stock[product]["cost"]
    price = stock[product]["price"]

    gross_margin = ((price - cost) / price) * 100

    print(f"Gross margin of {product} is {gross_margin:.2f}%")
```

## Task 6: Stall Performance Report

In this task, I will combine the results from the previous tasks into a formatted performance report using Python f-strings. The report will provide Wanjiku with a clear overview of sales performance and stock management.

```python
print("=" * 40)
print("       STALL PERFORMANCE REPORT")
print("=" * 40)

print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")

print("\n--- Sales Performance ---")
print(f"Highest Sales: KES {highest_sales:,} ({best_day}, Week {best_week})")
print(f"Lowest Sales: KES {lowest_sales:,} ({worst_day}, Week {worst_week})")
print(f"Week 1 Sales: KES {week1_total:,}")
print(f"Week 2 Sales: KES {week2_total:,}")
print(f"Week-on-Week Growth: {growth:.2f}%")

print("\n--- Daily Sales Classification ---")
print(f"Strong Days: {counts['Strong']}")
print(f"Normal Days: {counts['Normal']}")
print(f"Weak Days: {counts['Weak']}")

print("\n--- Weekday vs Weekend Performance ---")
print(f"Average Weekday Sales: KES {avg_weekday_sales:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend_sales:,.2f}")
print(f"Weekend Sales Advantage: {percentage_difference:.2f}%")

print("\n--- Stock Management ---")

print("Products to Reorder:", ", ".join(reorder_list(stock)))

print("\n--- Gross Margins ---")

for product in stock:
    cost = stock[product]["cost"]
    price = stock[product]["price"]
    gross_margin = ((price - cost) / price) * 100

    print(f"{product}: {gross_margin:.2f}%")

print("=" * 40)
```
